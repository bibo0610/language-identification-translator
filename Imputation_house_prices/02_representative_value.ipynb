### **02 - REPRESENTATIVE VALUE**

Tiếp tục sử dụng *LotFrontage*.
- Tính Mean của các giá trị hiện có.
- Tính Median.
- Thay thế các giá trị thiếu bằng Mean hoặc Median.
- So sánh kết quả.

#### 1. Đọc dữ liệu

```python
from pathlib import Path
import pandas as pd

DATA_PATH = Path("./data/train.csv")
df = pd.read_csv(DATA_PATH)

FEATURE = "LotFrontage"
SHOW_COLS = ["Id", FEATURE, "OverallQual", "SalePrice"]

print("So dong, so cot:", df.shape)
print("So gia tri thieu o LotFrontage:", df[FEATURE].isna().sum())
```

#### 2. Quan sát dữ liệu ban đầu

```python
missing_rows = df.index[df[FEATURE].isna()][:5]

print("5 dòng ban đâu có LotFrontage bị thiếu:")
display(df.loc[missing_rows, SHOW_COLS])

print("Tong so dong ban dau:", len(df))
print("So dong co LotFrontage bi thieu:", int(df[FEATURE].isna().sum()))
```

Ở 25 dòng đầu, *LotFrontage* có 5 giá trị *Nan* ở những dòng dữ liệu id = [8, 13, 15, 17, 25]

Tổng số dòng ở tập train.csv ban đầu là 1460 và có 259 dòng bị thiếu ở cột *LotFrontage*

#### 3. Điền giá trị thiếu bằng Mean

```python
# Tính mean
mean_value = df[FEATURE].mean()

print(f"Mean cua {FEATURE}: {mean_value:.4f}")
```

```python
df_mean = df.copy()
df_mean[FEATURE] = df_mean[FEATURE].fillna(mean_value)

print("Cung 5 dong do sau khi dien Mean:")
display(df_mean.loc[missing_rows, SHOW_COLS])
print("So gia tri thieu con lai:", int(df_mean[FEATURE].isna().sum()))
```

Có thể thấy, các dòng **NaN** đã được thay bằng giá trị **Mean = 70.0500** 

#### 4. Điền giá trị thiếu bằng Median

```python
# Tính median
median_value = df[FEATURE].median()

print(f"Median cua {FEATURE}: {median_value:.4f}")
```

```python
df_median = df.copy()
df_median[FEATURE] = df_median[FEATURE].fillna(median_value)

print("Cung 5 dong do sau khi dien Median:")
display(df_median.loc[missing_rows, SHOW_COLS])
print("So gia tri thieu con lai:", int(df_median[FEATURE].isna().sum()))
```

Các dòng bị thiếu đã được điền bằng giá trị của **Median**.

```python
output_dir = Path("output")
output_dir.mkdir(exist_ok=True)

mean_path = output_dir / "representative_value_mean.csv"
median_path = output_dir / "representative_value_median.csv"

df_mean.to_csv(mean_path, index=False)
df_median.to_csv(median_path, index=False)

print("Da luu:", mean_path.name)
print("Da luu:", median_path.name)
```


**Nhận xét:**

- **Mean:** Sử dụng giá trị trung bình của các quan sát hiện có để thay thế dữ liệu thiếu.
- **Median:** Sử dụng giá trị trung vị, ít bị ảnh hưởng bởi các giá trị ngoại lệ (Outliers) hơn Mean.

Đối với thuộc tính `LotFrontage`trong bài:
- Mean = 70.05
- Median = 69.00

Cả hai phương pháp đều giúp xử lý các giá trị thiếu trong `LotFrontage`. Tuy nhiên, việc lựa chọn Mean hay Median cần dựa trên đặc điểm phân phối dữ liệu và sự xuất hiện của các giá trị ngoại lệ.
