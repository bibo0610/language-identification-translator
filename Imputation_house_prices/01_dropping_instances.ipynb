### **01 - DROPPING INSTANCES**

Chọn thuộc tính *LotFrontage* (chiều dài mặt tiền lô đất), trong dataset có một số giá trị bị thiếu.
- Lọc các dòng có LotFrontage bị thiếu.
- Loại bỏ những dòng đó.
- So sánh số lượng mẫu trước và sau khi xóa.

#### 1. Đọc dữ liệu

```python
from pathlib import Path
import pandas as pd

DATA_PATH = Path("./data/train.csv")
df = pd.read_csv(DATA_PATH)

FEATURE = "LotFrontage"
SHOW_COLS = ["Id", FEATURE, "OverallQual", "SalePrice"]

print("So dong, so cot:", df.shape)
print("So gia tri thieu o LotFrontage:", df[FEATURE].isna().sum())
```

#### 2. Quan sát dữ liệu ban đầu

```python
print("Mot so dong du lieu dau tien cua train.csv ban dau:")
display(df[SHOW_COLS].head(20))

print("Tong so dong ban dau:", len(df))
print("So dong co LotFrontage bi thieu:", int(df[FEATURE].isna().sum()))
```

Ở 20 dòng đầu, *LotFrontage* có giá trị *Nan* ở những dòng dữ liệu id = [8, 13, 15, 17]

Tổng số dòng ở tập train.csv ban đầu là 1460 và có 259 dòng bị thiếu ở cột *LotFrontage*

#### 3. Áp dụng Dropping Instances

```python
df_dropped = df.dropna(subset=[FEATURE]).copy()

print("Sau khi ap dung Dropping Instances:")
display(df_dropped[SHOW_COLS].head(20))

print("So dong ban dau:", len(df))
print("So dong bi loai:", len(df) - len(df_dropped))
print("So dong con lai:", len(df_dropped))
print("So gia tri thieu con lai:", int(df_dropped[FEATURE].isna().sum()))
```

**Nhận xét:**

Ở đây, các dòng chứa giá trị thiếu có id = [8, 13, 15, 17] đã bị xóa.

Tập train.csv chỉ còn lại 1201 dòng vì đã xóa 259 dòng có chứa cột *LotFrontage = NaN*.

Phương pháp Dropping Instances giúp loại bỏ các giá trị thiếu một cách đơn giản. Tuy nhiên, việc xóa 259 dòng khiến tập dữ liệu giảm từ 1.460 xuống còn 1.201 dòng (mất khoảng 17,74% dữ liệu). Điều này có thể làm mất thông tin hữu ích và ảnh hưởng đến chất lượng dữ liệu.

```python
#Lưu kết quả
output_dir = Path("output")
output_dir.mkdir(exist_ok=True)
output_path = output_dir / "result_dropping_instances.csv"
df_dropped.to_csv(output_path, index=False)
print("Da luu file:", output_path.name)
```
